## Memory Usage

Check memory consumed by each column.

```python
memory_usage = df.memory_usage(deep=True)

memory_usage
```

```python
total_memory = memory_usage.sum() / (1024**2)

print(f"Total Memory Usage: {total_memory:.2f} MB")
```

# Missing Value Analysis

Before performing analysis, we need to identify missing values in every feature.
Missing values can affect statistics, visualizations, and machine learning models.

```python
# Missing values in each column

df.isnull().sum()
```

```python
# Percentage of missing values

missing_percentage = (
    df.isnull().sum() / len(df)
) * 100

missing_percentage.sort_values(ascending=False)
```

```python
# Missing Value Report

missing_report = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": missing_percentage.round(2)
})

missing_report = missing_report.sort_values(
    by="Missing Percentage",
    ascending=False
)

missing_report
```

```python
# Fill missing values

df["director"] = df["director"].fillna("Unknown")

df["cast"] = df["cast"].fillna("Not Available")

df["country"] = df["country"].fillna("Unknown")

# Fill rating using mode
df["rating"] = df["rating"].fillna(df["rating"].mode()[0])

# Fill duration using mode
df["duration"] = df["duration"].fillna(df["duration"].mode()[0])
```

```python
# Remove rows where date_added is missing

df = df.dropna(subset=["date_added"])
```

```python
# Verify cleaning

df.isnull().sum()
```

# Feature Engineering from Dates

Extract useful temporal features from date_added.

```python
# Create new features

df["year_added"] = df["date_added"].dt.year

df["month_added"] = df["date_added"].dt.month

df["month_name"] = df["date_added"].dt.month_name()

df["day_added"] = df["date_added"].dt.day
```

# Remove Extra Spaces

Text columns often contain unwanted leading or trailing spaces.

```python
# Remove spaces

text_columns = [
    "title",
    "director",
    "cast",
    "country",
    "rating",
    "duration",
    "genre"
]

for column in text_columns:
    df[column] = df[column].str.strip()
```

# Convert Pandas Columns to NumPy Arrays

NumPy performs mathematical operations on arrays faster than Python lists and is widely used in Machine Learning libraries.

```python
# Convert columns into NumPy arrays

release_year_array = df["release_year"].to_numpy()
duration_array = df["duration_numeric"].to_numpy()
```

```python
print(type(release_year_array))
print(type(duration_array))
```

## Mean of Release Year

Mean gives the average release year of Netflix content.

```python
mean_release_year = np.mean(release_year_array)
mean_duration = np.mean(duration_array)

print("Average Release Year :", round(mean_release_year, 2))
print("Average Duration :", round(mean_duration, 2))
```

## Median of Release Year

Median represents the middle value after sorting the data.

```python
median_release_year = np.median(release_year_array)
median_duration = np.median(duration_array)

print("Median Release Year :", median_release_year)
print("Median Duration :", median_duration)
```

## Mode of Rating

Mode is the most frequently occurring value.

```python
mode_rating = df["rating"].mode()[0]

print("Most Common Rating :", mode_rating)
```

## Minimum and Maximum Values

Find oldest and newest release years and shortest/longest durations.

```python
print("Oldest Release Year :", np.min(release_year_array))
print("Newest Release Year :", np.max(release_year_array))

print("Shortest Duration :", np.min(duration_array))
print("Longest Duration :", np.max(duration_array))
```

## Range

Range measures spread between maximum and minimum values.

```python
release_range = np.max(release_year_array) - np.min(release_year_array)
duration_range = np.max(duration_array) - np.min(duration_array)

print("Release Year Range :", release_range)
print("Duration Range :", duration_range)
```

## Variance

Variance measures how far values are spread from the mean.

```python
release_variance = np.var(release_year_array)
duration_variance = np.var(duration_array)

print("Variance (Release Year):", round(release_variance, 2))
print("Variance (Duration):", round(duration_variance, 2))
```

## Standard Deviation

Standard deviation is the square root of variance.

```python
release_std = np.std(release_year_array)
duration_std = np.std(duration_array)

print("Standard Deviation (Release Year):", round(release_std, 2))
print("Standard Deviation (Duration):", round(duration_std, 2))
```

## Percentile Analysis

Percentiles help understand data distribution.

```python
percentiles = np.percentile(
    duration_array,
    [25, 50, 75, 90, 95]
)

print(percentiles)
```

```python
#Quartiles individually

q1 = np.percentile(duration_array, 25)
q2 = np.percentile(duration_array, 50)
q3 = np.percentile(duration_array, 75)

print("Q1 :", q1)
print("Q2 :", q2)
print("Q3 :", q3)
```

## Interquartile Range (IQR)

IQR measures the spread of the middle 50% of the data.

```python
iqr = q3 - q1

print("Interquartile Range :", iqr)
```

## Outlier Detection Using IQR

```python
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = df[
    (df["duration_numeric"] < lower_bound) |
    (df["duration_numeric"] > upper_bound)
]

outliers.head()
```

## NumPy Statistical Summary

```python
statistics_summary = pd.DataFrame({
    "Metric": [
        "Mean",
        "Median",
        "Minimum",
        "Maximum",
        "Variance",
        "Standard Deviation"
    ],
    "Release Year": [
        np.mean(release_year_array),
        np.median(release_year_array),
        np.min(release_year_array),
        np.max(release_year_array),
        np.var(release_year_array),
        np.std(release_year_array)
    ],
    "Duration": [
        np.mean(duration_array),
        np.median(duration_array),
        np.min(duration_array),
        np.max(duration_array),
        np.var(duration_array),
        np.std(duration_array)
    ]
})

statistics_summary.round(2)
```
